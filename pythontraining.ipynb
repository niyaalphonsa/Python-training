```python
st="15,5,+,7,2,-,*"
st=st.split(",")
stack=[]
n=0
m=0
for i in st:
  if(i.isdigit()):
    stack.append(int(i))
  else:
    n=int(stack.pop())
    m=int(stack.pop())
    if(i=='-'):
      stack.append(m-n)
    elif(i=='+'):
      stack.append(m+n)
    elif(i=='*'):
      stack.append(m*n)
    elif(i=='/'):
      stack.append(m/n)
print(stack[-1])
```

```python
class node:
  def __init__(self,u):
    self.data=u
    self.left=None
    self.right=None

def height(root):
  if(root==None):
    return -1
  else:
    return 1+max(height(root.left),height(root.right))


root=node(10)
root.left=node(5)
root.right=node(15)
root.left.left=node(1)
root.left.right=node(7)
root.right.left=node(12)
height(root)
```
